# Análise Visual do TCC: Quadrantes da Produtividade Legislativa

Este notebook produz as figuras utilizadas no capítulo de **Resultados e Discussão** do TCC.
As visualizações incluem o gráfico de dispersão por quadrante, distribuição percentual,
destaques representativos, análise por partido e eficácia nas tramitações legislativas.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurações de estilo
sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

# Carregar dados
df = pd.read_csv('resultados.csv', sep=';', on_bad_lines='skip')
```

## 1. Gráfico de Dispersão dos Quadrantes

## Capítulo 4 – Resultados e Discussão

A aplicação do modelo aos dados da 57ª Legislatura resultou em uma base consolidada abrangendo todos os 513 deputados federais. A análise combinou produtividade legislativa ponderada e gastos com a cota parlamentar (CEAP), possibilitando a construção de rankings e de uma matriz de classificação por quadrantes.

A média de produtividade legislativa foi de 130,5 pontos (desvio padrão de 116,3), com variações entre 0 e 1.257 pontos. Já os gastos médios com CEAP foram de R$ 807 mil, com valores variando de R$ 15,19 a R$ 1,3 milhão. A mediana de R$ 872 mil indica uma distribuição levemente assimétrica.

```python
# Gráfico de dispersão: produtividade vs custo
plt.figure(figsize=(12,8))
ax = sns.scatterplot(
    data=df,
    x='gasto_ceap_ajustado',
    y='produtividade_legislativa',
    hue='quadrante',
    palette='Set2',
    alpha=0.7
)
plt.axhline(df['produtividade_legislativa'].median(), color='gray', linestyle='--')
plt.axvline(df['gasto_ceap_ajustado'].median(), color='gray', linestyle='--')
plt.title('Gráfico de Dispersão: Produtividade vs Custo do Mandato')
plt.xlabel('Gasto com CEAP (R$)')
plt.ylabel('Produtividade Legislativa (Pontos)')
plt.legend(title='Quadrante')
plt.tight_layout()
plt.savefig('dispersao_quadrantes.png')
plt.show()
```

A partir das medianas das variáveis, os deputados foram classificados da seguinte forma:

- Alta produtividade e baixo custo: 116 deputados
- Alta produtividade e alto custo: 140 deputados
- Baixa produtividade e baixo custo: 140 deputados
- Baixa produtividade e alto custo: 116 deputados

Essa distribuição equitativa reforça a adequação do critério de corte por mediana. Aproximadamente 22,6% dos parlamentares encontram-se no quadrante ideal (alto desempenho com baixo custo), enquanto outros 22,6% ocupam a faixa crítica de baixa entrega com alto gasto.

## 2. Distribuição Percentual por Quadrante

```python
dist_quadrantes = df['quadrante'].value_counts(normalize=True).sort_values(ascending=False)
dist_quadrantes.plot(kind='bar', color='skyblue')
plt.title('Distribuição Percentual por Quadrante')
plt.ylabel('Percentual')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('distribuicao_quadrantes.png')
plt.show()
```

## 3. Destaques Representativos (Top 5)

### Casos Representativos

**Destaque Positivo:** O deputado Amom Mandel (Cidadania – AM) lidera o ranking com 912,8 pontos e gastos ínfimos de R$ 15,19 com CEAP. Trata-se de um outlier positivo, com desempenho destacado e uso simbólico dos recursos.

**Eficiência Econômica:** Adriana Ventura (NOVO – SP), Kim Kataguiri (UNIÃO – SP) e Marcel van Hattem (NOVO – RS) combinam boa produtividade com gastos moderados, posicionando-se entre os mais eficientes da legislatura.

**Baixo Desempenho:** Foram identificados parlamentares com produtividade nula ou muito baixa, mesmo com gastos elevados. Parte desses casos envolve suplentes ou iniciantes no mandato, mas ainda assim exige atenção quanto à eficácia na aplicação de recursos públicos.

```python
top5 = df.sort_values(by='ranking').head(5)
sns.barplot(data=top5, y='nome', x='produtividade_legislativa', palette='viridis')
plt.title('Top 5 Deputados por Ranking Composto')
plt.xlabel('Produtividade Legislativa')
plt.ylabel('Deputado')
plt.tight_layout()
plt.savefig('top5_destaques.png')
plt.show()
```

## 4. Eficiência Proporcional por Partido (Top 10 em % no Quadrante I)

```python
partidos = df.groupby('sgPartido')['quadrante'].value_counts(normalize=True).unstack().fillna(0)
partidos_top = partidos.sort_values(by='Alta produtividade e baixo custo', ascending=False).head(10)
partidos_top['Alta produtividade e baixo custo'].plot(kind='bar', color='mediumseagreen')
plt.title('Proporção de Deputados no Quadrante I por Partido')
plt.ylabel('% no Quadrante I')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('eficiencia_partidos.png')
plt.show()
```

### Distribuição Proporcional por Quadrante (Empilhado por Partido)

O gráfico a seguir complementa a análise anterior ao mostrar, de forma empilhada,
a distribuição percentual de parlamentares por quadrante dentro dos 10 partidos
com maior percentual no Quadrante I. Essa visualização revela o perfil de eficiência
interna de cada legenda, permitindo identificar tanto partidos homogêneos quanto
aqueles com grande dispersão entre os quadrantes.

```python
# Gráfico de barras empilhadas por partido
partidos = df.groupby('sgPartido')['quadrante'].value_counts(normalize=True).unstack().fillna(0)
partidos_sorted = partidos.sort_values(by='Alta produtividade e baixo custo', ascending=False).head(10)

partidos_sorted.plot(
    kind='bar',
    stacked=True,
    color=['mediumseagreen', 'gold', 'skyblue', 'salmon']
)
plt.title('Distribuição Proporcional por Quadrante nos 10 Partidos com Mais Eficiência (Quadrante I)')
plt.ylabel('Percentual')
plt.xlabel('Partido')
plt.xticks(rotation=45)
plt.legend(title='Quadrante', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('quadrantes_partidos_empilhado.png')
plt.show()
```

### Tabela: Distribuição Percentual por Quadrante e Partido

A tabela abaixo mostra a proporção de deputados de cada partido alocados em cada quadrante. Os valores estão ordenados pelo percentual de parlamentares no Quadrante I (Alta produtividade e baixo custo).

```python
# Cálculo da distribuição percentual por quadrante e partido
tabela_quadrantes = df.groupby('sgPartido')['quadrante'].value_counts(normalize=True).unstack().fillna(0)
tabela_quadrantes = tabela_quadrantes.sort_values(by='Alta produtividade e baixo custo', ascending=False)
tabela_quadrantes_percent = (tabela_quadrantes * 100).round(1)
tabela_quadrantes_percent
```

### Eficiência Proporcional por Partido

Ao adotar uma análise proporcional, observou-se que:
- O PSOL apresenta 92% de seus deputados no Quadrante I (Alta Produtividade e Baixo Custo);
- Cidadania e PSB têm 50% e 53%, respectivamente, no Quadrante I;
- O NOVO, com apenas quatro parlamentares, tem 75% no Quadrante I.

Partidos maiores (PL, PT, MDB, PSD) apresentaram desempenho mais heterogêneo, refletindo a diversidade interna de suas bancadas.

## 5. Efetividade nas Tramitações Legislativas

### Temáticas Legislativas Predominantes

Entre os deputados mais produtivos, observaram-se atuações recorrentes nas seguintes áreas:
- Administração Pública
- Finanças e Orçamento
- Educação
- Direito Penal e Processual

A predominância dessas temáticas sugere alinhamento com pautas estruturantes e institucionalmente relevantes.

```python
df[['pct_sucesso', 'pct_fracasso', 'pct_andamento']].mean().plot(kind='bar', color=['green', 'red', 'gray'])
plt.title('Média de Situação das Proposições Legislativas')
plt.ylabel('Percentual')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('efetividade_tramitacoes.png')
plt.show()
```

### Efetividade nas Tramitações Legislativas

A análise dos desfechos das proposições revelou:
- Taxa média de sucesso (aprovação ou sanção): 0,3% a 0,5%;
- Fracassos (arquivadas, retiradas, rejeitadas): entre 20% e 30%;
- Maioria das proposições permanece em tramitação indefinida.

Deputados mais bem colocados no ranking tendem a apresentar:
- Maior taxa de proposições em estágio avançado;
- Menor taxa de fracasso;
- Leve vantagem na taxa de sucesso.

Esses padrões sugerem que desempenho legislativo de qualidade está associado à viabilidade institucional e à articulação política eficaz.
